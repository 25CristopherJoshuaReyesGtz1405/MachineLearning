#### **4.2 - Fase II - Preprocesamiento e Ingeniería de Características**
Habiendo validado empíricamente la influencia del género y la clase en la supervivencia, resulta imperativo adecuar la matriz de datos para el consumo algorítmico. Los clasificadores supervisados exigen matrices estrictamente numéricas y carentes de valores nulos (NaN). En consecuencia, esta fase ejecuta una depuración profunda: se descartan variables sin poder predictivo directo, se aplican transformaciones binarias a los atributos cualitativos y se emplea imputación estadística para reconstruir los expedientes incompletos.

##### **4.2.1 - Protocolos de Limpieza y Transformación Vectorial**

Para garantizar la viabilidad matemática de los modelos de *Machine Learning*, se aplican las siguientes reglas de negocio sobre el *DataFrame*:

1.  **Mapeo Categórico:** Se codifica la variable `Sex` bajo una nomenclatura binaria estricta donde `0 = masculino` y `1 = femenino`, tal como lo requiere la fundamentación del proyecto.
2.  **Depuración Estructural:** Se elimina la variable `Cabin` debido a que posee un déficit superior al 75% en sus registros, lo cual generaría ruido estadístico. Asimismo, se descartan columnas de identificación nominal (como `PassengerId`, `Name` y `Ticket`) que no aportan varianza geométrica medible.
3.  **Imputación Estadística:** Se resuelven los valores nulos en la variable `Age` mediante la inserción de la mediana aritmética (para mitigar el sesgo de datos atípicos), y se completan las ausencias menores en la variable `Embarked` utilizando la moda (el puerto de embarque más frecuente).
4.  **Codificación de Puertos:** Finalmente, la variable `Embarked` se transforma en vectores numéricos mediante la clase `LabelEncoder`, culminando así con un espacio multidimensional completamente limpio y listo para el entrenamiento computacional.

```python
# ==============================================================================
# 4.2.1 Ejecución de Limpieza e Ingeniería de Características
# ==============================================================================
from sklearn.preprocessing import LabelEncoder

# 1. Diagnóstico inicial de valores nulos
print("--- Presencia de Valores Nulos (Antes del Preprocesamiento) ---")
display(titanic_df.isnull().sum()[titanic_df.isnull().sum() > 0])

# 2. Mapeo binario de la variable 'Sex' (0 = Masculino, 1 = Femenino)
titanic_df['Sex'] = titanic_df['Sex'].map({'male': 0, 'female': 1})

# 3. Erradicación de columnas ruidosas y sin valor predictivo matemático
columnas_a_eliminar = ['Cabin', 'PassengerId', 'Name', 'Ticket']
titanic_df = titanic_df.drop(columns=columnas_a_eliminar, errors='ignore')

# 4. Imputación Estadística para datos faltantes
# 4.1 'Age': Se utiliza la mediana para evitar alteraciones por edades atípicas
mediana_edad = titanic_df['Age'].median()
titanic_df['Age'].fillna(mediana_edad, inplace=True)

# 4.2 'Embarked': Se utiliza la moda (el valor más frecuente)
moda_embarque = titanic_df['Embarked'].mode()[0]
titanic_df['Embarked'].fillna(moda_embarque, inplace=True)

# 4.3 'Fare': En caso de existir nulos aislados, se imputan con la mediana
titanic_df['Fare'].fillna(titanic_df['Fare'].median(), inplace=True)

# 5. Transformación numérica de la variable 'Embarked' (S, C, Q -> Numérico)
le_embarque = LabelEncoder()
titanic_df['Embarked'] = le_embarque.fit_transform(titanic_df['Embarked'])

# 6. Verificación de Integridad Post-Procesamiento
print("\n--- Integridad de la Matriz (Después del Preprocesamiento) ---")
print(titanic_df.info())

print("\nMatriz depurada y estructurada correctamente. Lista para partición y escalado.")
display(titanic_df.head())
```

##### **4.2.3 - Segmentación Poblacional y Estandarización de Escalas**

Previo al entrenamiento algorítmico, resulta indispensable dividir la matriz de información en dos subconjuntos independientes: uno destinado al aprendizaje y otro reservado para la validación. Siguiendo las directrices del proyecto, se destina el 70% de los registros al conjunto de entrenamiento y el 30% restante al conjunto de prueba. Es crucial destacar la aplicación del parámetro de estratificación (**stratify=y**), el cual garantiza que la proporción histórica de supervivientes y fallecidos se mantenga matemáticamente idéntica en ambas particiones, evitando así sesgos de muestreo.

Posterior a la partición, se ejecuta una transformación paramétrica mediante **StandardScaler**. Este procedimiento centra la media de cada variable en cero y escala su varianza a uno. Su aplicación es estrictamente obligatoria para modelos de naturaleza geométrica y de márgenes, como K-Nearest Neighbors (KNN) y Máquinas de Soporte Vectorial (SVM), puesto que estos algoritmos calculan distancias euclidianas. Sin esta estandarización, variables con magnitudes elevadas, como la tarifa (**Fare**), asfixiarían el peso matemático de variables críticas de menor escala, como la clase (**Pclass**) o el género (**Sex**). Cabe mencionar que, aunque algoritmos lógicos como los Árboles de Decisión no requieren este escalado, utilizar una matriz estandarizada globalmente garantiza un terreno de evaluación equitativo para el análisis comparativo posterior.

```python
# ==============================================================================
# 4.2.3 - Segmentación Poblacional y Estandarización de Escalas
# ==============================================================================
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# 1. Separación de la matriz predictora (X) y el vector objetivo (y)
X = titanic_df.drop('Survived', axis=1)
y = titanic_df['Survived']

# 2. Partición Estratégica (70% Entrenamiento - 30% Validación)
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.30,
    random_state=42,
    stratify=y  # Mantiene la proporción real de supervivencia
)

# 3. Instanciación del transformador estadístico
scaler = StandardScaler()

# 4. Ajuste y transformación sobre el conjunto de entrenamiento
X_train_scaled = scaler.fit_transform(X_train)

# 5. Transformación del conjunto de prueba (Prevención de Data Leakage)
X_test_scaled = scaler.transform(X_test)

print("Partición poblacional (70/30) y estandarización geométrica ejecutadas con éxito.")
print(f"Dimensiones de Entrenamiento: {X_train_scaled.shape}")
print(f"Dimensiones de Prueba: {X_test_scaled.shape}")
```

### **5. Fase III - Entrenamiento y Evaluación de Modelos Predictivos**

Con el espacio vectorial depurado y escalado, la investigación procede al despliegue computacional de las arquitecturas de clasificación. El propósito central de esta sección es someter el conjunto de entrenamiento a cinco paradigmas distintos de Inteligencia Artificial. Posteriormente, se evaluará la capacidad de generalización de cada modelo frente al conjunto de prueba, extrayendo las matrices de confusión y emitiendo un reporte exhaustivo de las métricas de rendimiento (Exactitud, Precisión, Sensibilidad y F1-Score).

#### **5.1 - Modelo 1: Regresión Logística (Enfoque Probabilístico Base)**

El flujo algorítmico da inicio con la implementación de la Regresión Logística, el cual opera como el estimador paramétrico de referencia. Este modelo probabilístico traza una relación lineal entre las características sociodemográficas y el logit de supervivencia, transformando el resultado mediante una función sigmoide para emitir una probabilidad entre 0 y 1. Su valor radica en la interpretabilidad y en su capacidad para establecer un punto de partida ("baseline") frente al rendimiento de modelos más complejos.

```python
# ==============================================================================
# 5.1 - Implementación y Evaluación: Regresión Logística
# ==============================================================================
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix

# 1. Instanciación del modelo paramétrico
lr_model = LogisticRegression(random_state=42)

# 2. Entrenamiento del algoritmo
print("Entrenando algoritmo de Regresión Logística...")
lr_model.fit(X_train_scaled, y_train)

# 3. Generación de predicciones espaciales y probabilísticas
y_pred_lr = lr_model.predict(X_test_scaled)
y_prob_lr = lr_model.predict_proba(X_test_scaled)[:, 1] # Probabilidades para futuro AUC

# 4. Extracción de Métricas de Rendimiento
print("\n--- Reporte de Clasificación: Regresión Logística ---")
print(classification_report(y_test, y_pred_lr, target_names=["No Sobrevivió (0)", "Sobrevivió (1)"]))

# 5. Representación Gráfica: Matriz de Confusión
matriz_lr = confusion_matrix(y_test, y_pred_lr)

plt.figure(figsize=(6, 4))
sns.heatmap(matriz_lr, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Sobrevivió', 'Sobrevivió'],
            yticklabels=['No Sobrevivió', 'Sobrevivió'],
            annot_kws={"size": 14, "weight": "bold"})

plt.title('Matriz de Confusión: Regresión Logística', fontsize=13, weight='bold', pad=12)
plt.xlabel('Predicción del Modelo', fontsize=11)
plt.ylabel('Estado Histórico Real', fontsize=11)
plt.tight_layout()
plt.show()
```

##### **5.1.1 - Optimización Paramétrica de la Regresión Logística**

Si bien el estimador paramétrico base ha demostrado una capacidad predictiva inicial sólida, resulta imperativo maximizar su eficiencia analítica para reducir la tasa de falsos negativos. Para lograr este cometido, se emplea la técnica de Búsqueda en Malla (*GridSearchCV*). Este procedimiento evalúa exhaustivamente múltiples combinaciones de hiperparámetros, tales como la fuerza de regularización inversa (**C**), la norma de penalización (**penalty**), el algoritmo de resolución (**solver**) y el balanceo de pesos poblacionales (**class_weight**). Mediante validación cruzada estratificada, el sistema identificará de manera autónoma la configuración matemática que ofrezca el mejor rendimiento y capacidad de generalización.

```python
# ==============================================================================
# 5.1.1 - Optimización de Hiperparámetros (GridSearchCV)
# ==============================================================================
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression

# 1. Definición del espacio de búsqueda paramétrica
param_grid_lr = {
    'C': [0.01, 0.1, 1, 10, 100],               # Fuerza de regularización
    'penalty': ['l1', 'l2'],                    # Tipo de penalización matemática
    'solver': ['liblinear', 'saga'],            # Optimizadores compatibles con l1 y l2
    'class_weight': [None, 'balanced']          # Tratamiento del desequilibrio de clases
}

# 2. Instanciación del motor de búsqueda con validación cruzada (cv=5)
# Se utiliza accuracy como métrica base, aunque el balanceo ayudará al recall
grid_search_lr = GridSearchCV(
    estimator=LogisticRegression(random_state=42),
    param_grid=param_grid_lr,
    cv=5,
    scoring='accuracy',
    n_jobs=-1,
    verbose=1
)

# 3. Ejecución del entrenamiento exhaustivo
print("Ejecutando optimización paramétrica para Regresión Logística...\n")
grid_search_lr.fit(X_train, y_train)

# 4. Extracción del modelo óptimo
best_lr_model = grid_search_lr.best_estimator_

print("\n--- Búsqueda Finalizada ---")
print(f"Mejor configuración matemática identificada: {grid_search_lr.best_params_}")

# 5. Evaluación del Modelo Optimizado
y_pred_lr_opt = best_lr_model.predict(X_test_scaled)
y_prob_lr_opt = best_lr_model.predict_proba(X_test_scaled)[:, 1] # Actualización para AUC

print("\n--- Reporte de Clasificación: Regresión Logística (Optimizada) ---")
print(classification_report(y_test, y_pred_lr_opt, target_names=["No Sobrevivió (0)", "Sobrevivió (1)"]))

# 6. Representación Gráfica: Matriz de Confusión Optimizada
matriz_lr_opt = confusion_matrix(y_test, y_pred_lr_opt)

plt.figure(figsize=(6, 4))
sns.heatmap(matriz_lr_opt, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Sobrevivió', 'Sobrevivió'],
            yticklabels=['No Sobrevivió', 'Sobrevivió'],
            annot_kws={"size": 14, "weight": "bold"})

plt.title('Matriz de Confusión: Regresión Logística (Optimizada)', fontsize=13, weight='bold', pad=12)
plt.xlabel('Predicción del Modelo', fontsize=11)
plt.ylabel('Estado Histórico Real', fontsize=11)
plt.tight_layout()
plt.show()
```

#### **5.2 - Modelo 2: K-Nearest Neighbors (Enfoque Geométrico)**

El segundo paradigma a evaluar corresponde a K-Nearest Neighbors (KNN). A diferencia de los modelos paramétricos que construyen una ecuación global, KNN es un algoritmo "perezoso" (*lazy learning*) fundamentado en la proximidad espacial. Su mecanismo de decisión consiste en proyectar a un pasajero en el espacio multidimensional y predecir su supervivencia basándose en el destino mayoritario de sus *k* vecinos más cercanos.

Es en esta arquitectura donde la estandarización paramétrica (**StandardScaler**), ejecutada en la fase de preprocesamiento, demuestra su valor crítico. Al unificar las escalas de todas las variables, se garantiza que el cálculo de la distancia euclidiana no sea sesgado por atributos con magnitudes elevadas (como la tarifa del boleto), permitiendo que factores determinantes como el género o la clase ejerzan su verdadera influencia geométrica.

```python
# ==============================================================================
# 5.2 - Implementación y Evaluación: K-Nearest Neighbors (KNN)
# ==============================================================================
from sklearn.neighbors import KNeighborsClassifier

# 1. Instanciación del modelo geométrico
# Se establece k=5 como parámetro estándar inicial
knn_model = KNeighborsClassifier(n_neighbors=5, metric='minkowski', p=2)

# 2. Ingesta espacial y mapeo de distancias (Entrenamiento)
print("Entrenando algoritmo K-Nearest Neighbors (k=5)...")
knn_model.fit(X_train_scaled, y_train)

# 3. Generación de predicciones espaciales y probabilísticas
y_pred_knn = knn_model.predict(X_test_scaled)
y_prob_knn = knn_model.predict_proba(X_test_scaled)[:, 1]

# 4. Extracción de Métricas de Rendimiento
print("\n--- Reporte de Clasificación: K-Nearest Neighbors ---")
print(classification_report(y_test, y_pred_knn, target_names=["No Sobrevivió (0)", "Sobrevivió (1)"]))

# 5. Representación Gráfica: Matriz de Confusión
matriz_knn = confusion_matrix(y_test, y_pred_knn)

plt.figure(figsize=(6, 4))
sns.heatmap(matriz_knn, annot=True, fmt='d', cmap='Greens',
            xticklabels=['No Sobrevivió', 'Sobrevivió'],
            yticklabels=['No Sobrevivió', 'Sobrevivió'],
            annot_kws={"size": 14, "weight": "bold"})

plt.title('Matriz de Confusión: K-Nearest Neighbors', fontsize=13, weight='bold', pad=12)
plt.xlabel('Predicción del Modelo', fontsize=11)
plt.ylabel('Estado Histórico Real', fontsize=11)
plt.tight_layout()
plt.show()
```

#### **5.3 - Modelo 3: Árboles de Decisión (Enfoque Lógico-Deductivo)**

El tercer estimador implementado introduce un paradigma radicalmente distinto: el aprendizaje basado en reglas lógicas. Los Árboles de Decisión emulan el razonamiento deductivo humano al particionar el conjunto de datos mediante preguntas jerárquicas (por ejemplo, dividiendo inicialmente a la población por género y posteriormente por clase o edad).

A diferencia de KNN o SVM, los árboles de decisión son insensibles a la magnitud de las escalas, por lo que podrían operar con la matriz cruda; sin embargo, se emplea el conjunto estandarizado para mantener la rigurosidad y equidad en la comparativa final. Cabe destacar que la principal vulnerabilidad de este algoritmo es su propensión al sobreajuste (*overfitting*), motivo por el cual se instanciará con un límite de profundidad (**max_depth**) que actúe como medida de regularización estructural, forzando al modelo a generalizar patrones en lugar de memorizar el ruido estadístico.

```python
# ==============================================================================
# 5.3 - Implementación y Evaluación: Árboles de Decisión
# ==============================================================================
from sklearn.tree import DecisionTreeClassifier

# 1. Instanciación del modelo lógico
# Se aplica un max_depth=5 para prevenir el sobreajuste masivo en las ramas
dt_model = DecisionTreeClassifier(max_depth=5, criterion='gini', random_state=42)

# 2. Construcción de la jerarquía de reglas (Entrenamiento)
print("Entrenando algoritmo de Árbol de Decisión...")
dt_model.fit(X_train_scaled, y_train)

# 3. Generación de predicciones lógicas y probabilísticas
y_pred_dt = dt_model.predict(X_test_scaled)
y_prob_dt = dt_model.predict_proba(X_test_scaled)[:, 1]

# 4. Extracción de Métricas de Rendimiento
print("\n--- Reporte de Clasificación: Árbol de Decisión ---")
print(classification_report(y_test, y_pred_dt, target_names=["No Sobrevivió (0)", "Sobrevivió (1)"]))

# 5. Representación Gráfica: Matriz de Confusión
matriz_dt = confusion_matrix(y_test, y_pred_dt)

plt.figure(figsize=(6, 4))
sns.heatmap(matriz_dt, annot=True, fmt='d', cmap='Oranges',
            xticklabels=['No Sobrevivió', 'Sobrevivió'],
            yticklabels=['No Sobrevivió', 'Sobrevivió'],
            annot_kws={"size": 14, "weight": "bold"})

plt.title('Matriz de Confusión: Árbol de Decisión', fontsize=13, weight='bold', pad=12)
plt.xlabel('Predicción del Modelo', fontsize=11)
plt.ylabel('Estado Histórico Real', fontsize=11)
plt.tight_layout()
plt.show()
```

#### **5.4 - Modelo 4: Random Forest (Enfoque de Ensamble)**

Con el propósito de mitigar la limitación del Árbol de Decisión respecto al balance entre precisión y sensibilidad (Recall), se implementa la arquitectura *Random Forest*. Este algoritmo representa el pináculo del paradigma de ensamble (*Bagging*); en lugar de depender de una única estructura lógica, genera un "bosque" compuesto por múltiples árboles independientes, entrenados sobre subconjuntos aleatorios de los datos.

La predicción final se obtiene mediante un mecanismo de votación mayoritaria. Esta diversificación algorítmica reduce drásticamente la varianza del modelo y suprime el riesgo de sobreajuste, logrando capturar patrones complejos con una robustez significativamente mayor frente a la presencia de ruido estadístico en los registros de supervivencia.

```python
# ==============================================================================
# 5.4 - Implementación y Evaluación: Random Forest
# ==============================================================================
from sklearn.ensemble import RandomForestClassifier

# 1. Instanciación del modelo de ensamble
# Se configuran 100 estimadores (árboles) con un límite de profundidad controlado
rf_model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=42)

# 2. Entrenamiento del bosque (Ingesta de datos)
print("Entrenando algoritmo de ensamble Random Forest...")
rf_model.fit(X_train_scaled, y_train)

# 3. Generación de predicciones por votación mayoritaria y cálculo probabilístico
y_pred_rf = rf_model.predict(X_test_scaled)
y_prob_rf = rf_model.predict_proba(X_test_scaled)[:, 1]

# 4. Extracción de Métricas de Rendimiento
print("\n--- Reporte de Clasificación: Random Forest ---")
print(classification_report(y_test, y_pred_rf, target_names=["No Sobrevivió (0)", "Sobrevivió (1)"]))

# 5. Representación Gráfica: Matriz de Confusión
matriz_rf = confusion_matrix(y_test, y_pred_rf)

plt.figure(figsize=(6, 4))
sns.heatmap(matriz_rf, annot=True, fmt='d', cmap='Purples',
            xticklabels=['No Sobrevivió', 'Sobrevivió'],
            yticklabels=['No Sobrevivió', 'Sobrevivió'],
            annot_kws={"size": 14, "weight": "bold"})

plt.title('Matriz de Confusión: Random Forest', fontsize=13, weight='bold', pad=12)
plt.xlabel('Predicción del Modelo', fontsize=11)
plt.ylabel('Estado Histórico Real', fontsize=11)
plt.tight_layout()
plt.show()
```

#### **5.5 - Modelo 5: Máquinas de Soporte Vectorial (SVM)**

El cierre de la etapa de modelado se confía a las Máquinas de Soporte Vectorial (SVM), un estimador reconocido por su potencia en la clasificación de alta complejidad geométrica. A diferencia de los métodos anteriores, SVM busca trazar un "hiperplano óptimo" en el espacio multidimensional que logre la máxima separación posible (margen) entre las clases de sobrevivientes y no sobrevivientes.

La correcta ejecución de este algoritmo exige rigurosamente la estandarización de las escalas implementada en la sección metodológica, previniendo distorsiones en el cálculo de los márgenes. Asimismo, resulta imperativo activar el cálculo de probabilidades (`probability=True`) durante la instanciación del objeto, permitiendo así su integración en la comparativa posterior mediante el Área Bajo la Curva (AUC).

```python
# ==============================================================================
# 5.5 - Implementación y Evaluación: Máquinas de Soporte Vectorial (SVM)
# ==============================================================================
from sklearn.svm import SVC

# 1. Instanciación del modelo SVM
# Se habilita 'probability=True' explícitamente para permitir el cálculo de la curva ROC-AUC
svm_model = SVC(kernel='rbf', probability=True, random_state=42)

# 2. Búsqueda del hiperplano óptimo (Entrenamiento)
print("Entrenando algoritmo de Máquinas de Soporte Vectorial (SVM)...")
svm_model.fit(X_train_scaled, y_train)

# 3. Generación de predicciones y probabilidades
y_pred_svm = svm_model.predict(X_test_scaled)
y_prob_svm = svm_model.predict_proba(X_test_scaled)[:, 1]

# 4. Extracción de Métricas de Rendimiento
print("\n--- Reporte de Clasificación: SVM ---")
print(classification_report(y_test, y_pred_svm, target_names=["No Sobrevivió (0)", "Sobrevivió (1)"]))

# 5. Representación Gráfica: Matriz de Confusión
matriz_svm = confusion_matrix(y_test, y_pred_svm)

plt.figure(figsize=(6, 4))
sns.heatmap(matriz_svm, annot=True, fmt='d', cmap='Reds',
            xticklabels=['No Sobrevivió', 'Sobrevivió'],
            yticklabels=['No Sobrevivió', 'Sobrevivió'],
            annot_kws={"size": 14, "weight": "bold"})

plt.title('Matriz de Confusión: SVM', fontsize=13, weight='bold', pad=12)
plt.xlabel('Predicción del Modelo', fontsize=11)
plt.ylabel('Estado Histórico Real', fontsize=11)
plt.tight_layout()
plt.show()
```

### **Fase IV - Interpretación Individual de Resultados**

#### **6.1 - Análisis del Modelo 1 - El Algoritmo De La Regresión Logística**

Iniciando con la evaluación individual de los clasificadores, la Regresión Logística opera como un algoritmo paramétrico que establece una relación lineal entre las variables independientes y el logaritmo de las probabilidades, procesando el resultado mediante una función sigmoide para emitir una predicción binaria.

Entre sus principales ventajas se encuentra su alta interpretabilidad y eficiencia computacional; sin embargo, su limitación radica en la incapacidad de modelar fronteras de decisión complejas o no lineales sin la creación previa de variables polinomiales. Al analizar sus resultados, se observa un desempeño sumamente equilibrado, logrando detectar de manera efectiva tanto a los no sobrevivientes, con una exactitud del 82%, como a los sobrevivientes, alcanzando una sensibilidad del 71%.

Las posibles causas de este rendimiento competitivo se fundamentan en la fuerte correlación lineal que variables como el género y la clase social ejercen de manera natural sobre el destino de los pasajeros. Para optimizar aún más este estimador, las recomendaciones técnicas sugieren implementar la ingeniería de características mediante la creación de términos de interacción matemática, así como un ajuste minucioso del parámetro de regularización para calibrar el margen de error tolerado.

#### **6.2 - Análisis del Modelo 2 - El Algoritmo De K-Nearest Neighbors (KNN)**

Por otro lado, el algoritmo K-Nearest Neighbors (KNN) aborda el problema predictivo bajo un principio geométrico, evaluando la proximidad espacial mediante el cálculo de distancias euclidianas para clasificar a un pasajero según el destino mayoritario de sus vecinos más cercanos. Aunque esta naturaleza basada en instancias le otorga la ventaja de adaptarse a fronteras no lineales sin una estructura matemática predefinida, presenta la gran limitación de ser extremadamente sensible a las diferencias de escala y al ruido estadístico.

Respecto a su interpretación de resultados, el modelo agrupa con gran exactitud a los pasajeros fallecidos, pero presenta una disminución en la detección de sobrevivientes reales, registrando una sensibilidad del 68%. La causa principal de esta omisión de casos positivos es el solapamiento espacial en el conjunto de datos, donde pasajeros con perfiles sociodemográficos similares experimentaron destinos opuestos por factores aleatorios durante la evacuación.

Con el propósito de mitigar este efecto y mejorar el modelo, se recomienda ejecutar una validación cruzada exhaustiva para determinar el valor óptimo del número de vecinos, además de explorar métricas de distancia alternativas que ofrezcan mayor robustez.

#### **6.3 - Análisis del Modelo 3 - Árbol de Decisión**

En contraste con los enfoques anteriores, el Árbol de Decisión construye una jerarquía deductiva basada en reglas lógicas, dividiendo iterativamente a la población mediante la evaluación de la reducción de impureza de cada variable. Su principal fortaleza es la transparencia absoluta que ofrece en la toma de decisiones; no obstante, es altamente propenso al sobreajuste si se le permite ramificarse sin restricciones.

Al interpretar sus métricas, se detecta el comportamiento más conservador de la investigación: resulta ser el estimador más eficaz para identificar correctamente a los no sobrevivientes, pero omite a casi la mitad de los sobrevivientes reales, arrojando una sensibilidad de apenas el 59%, aunque respaldada por una elevada precisión del 80% en los escasos dictámenes positivos que emite.

Este rendimiento asimétrico es consecuencia directa de la regularización impuesta a su profundidad máxima, la cual forzó al modelo a simplificar sus reglas y priorizar la clase mayoritaria. Para perfeccionar su estructura, se recomienda afinar hiperparámetros de segmentación interna y aplicar una estricta selección de características, eliminando variables que introduzcan divisiones lógicas redundantes.

#### **6.4 - Análisis del Modelo 4 - El Random Forest**

Para contrarrestar la inestabilidad de la arquitectura lógica individual, el modelo Random Forest implementa un paradigma de ensamble, combinando cientos de árboles de decisión entrenados sobre submuestras aleatorias para emitir un veredicto mediante votación mayoritaria.

Esta diversificación algorítmica posee la enorme ventaja de reducir drásticamente la varianza, logrando romper el límite predictivo de los algoritmos base para alcanzar una exactitud global del 81%; sin embargo, su complejidad transforma al modelo en un sistema opaco, dificultando la interpretación precisa de sus reglas internas. Durante la evaluación de sus predicciones, el bosque aleatorio demuestra un equilibrio sólido al detectar a los no sobrevivientes y asegura la mayor precisión del estudio (81%) al predecir a los sobrevivientes, aunque mantiene una sensibilidad moderada del 66%.

Las causas de este éxito estadístico radican en la promediación de decisiones, la cual filtra el ruido individual de cada árbol y consolida los patrones más fuertes. Como escenario de mejora, se plantea incrementar la cantidad de estimadores y calibrar la proporción de variables consideradas en cada partición, fomentando así una mayor exploración de interacciones complejas.

#### **6.5 - Análisis del Modelo 5 - Máquinas de Soporte Vectorial (SVM)**

Finalmente, la evaluación del modelo SVM —configurado con un *kernel* de base radial (RBF)— corrobora su reputación como un clasificador altamente robusto en espacios no lineales. Al igual que el Random Forest, alcanza una exactitud general del **81%**, pero lo hace exhibiendo una armonía predictiva superior.

El modelo reporta una **Precisión de 0.78** y un **Recall de 0.71**, evidenciando una notable capacidad para trazar fronteras de decisión complejas que equilibran la sensibilidad y la exactitud. Gracias a la estandarización paramétrica previa, el SVM logra maximizar el margen de separación entre ambas clases, posicionándose, junto a la regresión logística, como una de las alternativas más estables al no presentar caídas drásticas en su capacidad de detección de verdaderos positivos.

### **7 - La Comparación Entre Los Resultados Obtenidos**

#### **7.1 - Análisis Comparativo**
Tras la evaluación individual de las arquitecturas de aprendizaje supervisado, resulta imperativo ejecutar un análisis comparativo que permita identificar el modelo con mayor capacidad predictiva y equilibrio estadístico. Esta fase consolida las métricas de Exactitud, Sensibilidad (Recall) para ambas clases, F1-Score y el Área Bajo la Curva (AUC) en una matriz comparativa. Adicionalmente, se integra el análisis visual mediante el trazado de las Curvas ROC (Receiver Operating Characteristic), las cuales exponen la relación dinámica entre la tasa de verdaderos positivos y falsos positivos, facilitando la determinación de la arquitectura más robusta ante la complejidad de los datos del Titanic.

```python
# ==============================================================================
# 7.1 - Consolidación de Métricas y Tabla Resumen
# ==============================================================================
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, recall_score, f1_score, roc_auc_score, roc_curve

# 1. Definición de la estructura de datos para la comparativa
modelos_nombres = ['Regresión Logística', 'KNN', 'Árbol de Decisión', 'Random Forest', 'SVM']
predicciones = [y_pred_lr_opt, y_pred_knn, y_pred_dt, y_pred_rf, y_pred_svm]
probabilidades = [y_prob_lr_opt, y_prob_knn, y_prob_dt, y_prob_rf, y_prob_svm]

datos_comparativos = []

for i in range(len(modelos_nombres)):
    acc = accuracy_score(y_test, predicciones[i])
    rec_0 = recall_score(y_test, predicciones[i], pos_label=0) # Recall No Sobrevivió
    rec_1 = recall_score(y_test, predicciones[i], pos_label=1) # Recall Sobrevivió
    f1 = f1_score(y_test, predicciones[i])
    auc = roc_auc_score(y_test, probabilidades[i])

    comentario = ""
    if modelos_nombres[i] == 'Random Forest': comentario = "Mayor exactitud global."
    elif modelos_nombres[i] == 'Regresión Logística': comentario = "Modelo más equilibrado (Recall)."
    elif modelos_nombres[i] == 'SVM': comentario = "Alta robustez en fronteras."
    elif modelos_nombres[i] == 'Árbol de Decisión': comentario = "Sesgado a la clase mayoritaria."
    else: comentario = "Sensible al ruido espacial."

    datos_comparativos.append([modelos_nombres[i], acc, rec_0, rec_1, f1, auc, comentario])

# 2. Creación del DataFrame resumen
df_resumen = pd.DataFrame(datos_comparativos, columns=[
    'Modelo', 'Accuracy', 'Recall (No Sobv.)', 'Recall (Sobv.)', 'F1-score', 'AUC', 'Comentario'
])

print("--- Tabla Resumen de Desempeño Comparativo ---")
display(df_resumen.style.background_gradient(cmap='Blues', subset=['Accuracy', 'F1-score', 'AUC']))
print("\n")
# ==============================================================================
# 7.2 - Visualización Gráfica: Comparativa General y Curvas ROC
# ==============================================================================
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Gráfico A: Barras Comparativas de Accuracy, F1-Score y AUC
df_plot = df_resumen.melt(id_vars='Modelo', value_vars=['Accuracy', 'F1-score', 'AUC'],
                          var_name='Métrica', value_name='Valor')

sns.barplot(data=df_plot, x='Modelo', y='Valor', hue='Métrica', palette='viridis', ax=ax1)
ax1.set_title('Métricas de Rendimiento por Arquitectura', fontsize=14, weight='bold', pad=15)
ax1.set_ylim(0.6, 0.9)
ax1.grid(axis='y', linestyle='--', alpha=0.7)
ax1.tick_params(axis='x', rotation=15)

# Gráfico B: Análisis de Curvas ROC Multiclasificador
colores_roc = ['#e74c3c', '#2ecc71', '#f39c12', '#9b59b6', '#34495e']

for i in range(len(modelos_nombres)):
    fpr, tpr, _ = roc_curve(y_test, probabilidades[i])
    ax2.plot(fpr, tpr, color=colores_roc[i], lw=2,
             label=f'{modelos_nombres[i]} (AUC = {df_resumen["AUC"].iloc[i]:.3f})')

# Línea base de azar (modelo aleatorio)
ax2.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--')
ax2.set_xlim([0.0, 1.0])
ax2.set_ylim([0.0, 1.05])
ax2.set_xlabel('Tasa de Falsos Positivos (1 - Especificidad)', fontsize=12)
ax2.set_ylabel('Tasa de Verdaderos Positivos (Sensibilidad)', fontsize=12)
ax2.set_title('Curvas ROC Multiclasificador', fontsize=14, weight='bold', pad=15)
ax2.legend(loc="lower right", fontsize=11)
ax2.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

#### **7.2 - Análisis Comparativo Final y Conclusión**

Tras el escrutinio sistémico de la matriz de resultados y la evaluación geométrica de las curvas ROC, se determina que el modelo que mejor predice la supervivencia es el ensamble **Random Forest**. Esta arquitectura logra el punto de optimización más alto al registrar la mayor exactitud global (81%) y dominar el Área Bajo la Curva (AUC), evidenciando una separación de clases superior a la de los métodos individuales.

No obstante, es imperativo establecer una distinción técnica fundamental: si bien el ensamble maximiza la precisión general, la **Regresión Logística** demuestra una estabilidad analítica sobresaliente al ofrecer el mejor balance entre la sensibilidad (*Recall*) de ambas categorías poblacionales. Esta armonía probabilística es de vital importancia cuando el objetivo de la investigación exige identificar correctamente tanto a los sobrevivientes como a los fallecidos sin incurrir en un sesgo perjudicial hacia la clase dominante.

En conclusión, el fenómeno de supervivencia abordado posee patrones inherentemente no lineales que benefician al ensamble, aunque los estimadores paramétricos permanecen como el estándar innegable para aplicaciones que priorizan la simetría métrica y la interpretabilidad directa.

### **9 - Reflexión Final y Análisis Sociodemográfico**
Más allá de la optimización matemática de los hiperplanos y las matrices de confusión, el verdadero valor de la Inteligencia Artificial radica en su capacidad para develar la naturaleza del comportamiento humano bajo condiciones extremas. Las métricas extraídas en esta investigación trascienden la estadística, constituyendo un registro histórico de las dinámicas sociales imperantes a principios del siglo XX.


##### **9.1 - ¿Qué modelo funcionó mejor y por qué?**
Desde una perspectiva estrictamente cuantitativa, el ensamble **Random Forest** demostró ser la arquitectura superior, alcanzando la mayor exactitud global (81%) y el Área Bajo la Curva (AUC) más robusta. Este éxito se atribuye a su mecanismo de votación mayoritaria, el cual logró promediar múltiples árboles de decisión para filtrar el "ruido" estadístico generado por el caos propio del naufragio, enfocándose exclusivamente en los predictores más fuertes. Sin embargo, para aplicaciones que demandan un equilibrio crítico entre la sensibilidad de fallecidos y sobrevivientes sin sesgarse por el volumen de la clase mayoritaria, la **Regresión Logística** probó ser el modelo más estable y armónico del estudio.

##### **9.2 - ¿Qué variables influyen más en el destino?**
El análisis empírico confirma categóricamente que la supervivencia en el RMS Titanic no estuvo dictada por el azar, sino por una estricta jerarquización basada en dos variables fundamentales:
1.  **El Género (`Sex`):** Constituye el vector discriminante más poderoso del modelo. Ser mujer incrementaba exponencialmente las probabilidades matemáticas de sobrevivir.
2.  **La Clase Socioeconómica (`Pclass`):** El poder adquisitivo, reflejado en la tarifa del boleto (`Fare`) y la clase de la cabina, actuó como la segunda frontera de decisión más determinante, trazando una clara división en la tasa de mortalidad poblacional.

##### **9.3 - ¿Qué patrones sociales se observan?**
Los algoritmos de aprendizaje automático lograron decodificar e ilustrar dos fenómenos sociológicos que definieron el siniestro:
* **La Ejecución del Protocolo "Mujeres y Niños Primero":** La alta sensibilidad de los modelos al separar los géneros valida históricamente la aplicación del código de conducta marítima de Birkenhead. El modelo matemático asimiló que, independientemente de otros factores, el sacrificio sistemático de los pasajeros masculinos fue la norma social imperante durante el proceso de evacuación.
* **La Estratificación Social Eduardiana:** La asimetría en la supervivencia entre clases revela un trágico patrón logístico y de clase. Los pasajeros de Primera Clase, debido a la ubicación privilegiada de sus camarotes en las cubiertas superiores y su cercanía directa a la zona de botes salvavidas, gozaron de una ventaja de supervivencia abrumadora frente a los contingentes de Tercera Clase, quienes se encontraban confinados en los niveles inferiores de la embarcación con barreras físicas y lingüísticas severas.

En conclusión, este proyecto de Machine Learning demuestra que los modelos clasificadores no solo han predicho una etiqueta binaria (1 o 0), sino que han logrado parametrizar, con un 81% de exactitud, la cruda realidad de un evento histórico donde la supervivencia fue un privilegio estructurado por el estatus socioeconómico y el género.
